## 10-30: Lists keep an ordered sequence
A list stores several values between square brackets, separated by commas. It preserves order and allows duplicates. Access by zero-based index; slice with excluded stop, like a string. Lists are mutable: you can replace an item. A list of synthetic DNA strings is useful when order matters and items may change. Expect AT, then ['AT', 'GC'], then ['AT', 'GC', 'AT'] below.

````python
sequences = ["AT", "GC", "AT"]
print(sequences[0])
print(sequences[:2])
print(sequences)
````

`append(value)` adds one item at the end. `remove(value)` removes the first equal item; it does not take an index. If the value is absent, remove raises ValueError. These methods change the list in place; do not write `sequences = sequences.append(...)`, because append returns None (no result value). `len` counts list items. `+` joins lists into a new list. Expect ['ATAT', 'AT', 'TT'], length 3, then a new list with CC added. Practice 1-2.

````python
sequences = ["AT", "GC", "AT"]
sequences[0] = "ATAT"
sequences.append("TT")
sequences.remove("GC")
print(sequences)
print(len(sequences))
print(sequences + ["CC"])
````

## 30-50: Tuples, dictionaries and sets, one at a time
A tuple stores an ordered sequence using commas, usually in parentheses. It allows duplicates and indexing, but it is immutable: its stored references cannot be replaced, appended or removed. Use it for a fixed record. Here a pair of fictional lengths in mm has first=10.0 and second=12.0. A one-item tuple needs a comma: `(10.0,)`; `(10.0)` alone is a float. These tuples contain numbers, so there is no mutable item inside. A tuple can contain a mutable list, so immutable does not mean everything nested inside is frozen.

````python
length_pair_mm = (10.0, 12.0)
print(length_pair_mm[0])
print(length_pair_mm[1])
print(len(length_pair_mm))
````

A dictionary maps unique keys to values: `{key: value, ...}`. Use a sequence identifier (a unique label) as a key and DNA text as its value. Brackets look up a key, not a position. Modern Python dictionaries preserve insertion order; assigning an existing key replaces its value, not a second record. Duplicate literal keys also keep only the last value. Use unique identifiers before building the dictionary or a record can be lost. Accessing an absent key raises KeyError. Expect GC, then {'One': 'AT', 'Two': 'GCGC', 'Three': 'TT'}.

````python
dna_by_id = {"One": "AT", "Two": "GC"}
print(dna_by_id["Two"])
dna_by_id["Two"] = "GCGC"
dna_by_id["Three"] = "TT"
print(dna_by_id)
````

A set stores unique hashable elements. For today's examples, strings and numbers can be set elements or dictionary keys; lists and dictionaries cannot, because they are mutable and unhashable. Tuples of hashable items can be keys/elements. Sets allow membership tests but have no positional index and promise no iteration/display order. Use a set for unique DNA symbols or identifiers. `set("AATG")` visits characters and keeps A,T,G once each. `set()` is an empty set; `{}` is an empty dictionary. `add(value)` adds a set member; adding an existing member changes nothing. To remove an existing member, use remove(value); an absent member raises KeyError. Expect 3, then True, then 3. Practice 3-5.

````python
symbols = set("AATG")
print(len(symbols))
symbols.add("C")
print("C" in symbols)
symbols.remove("C")
print(len(symbols))
````

## 50-65: Mutability, membership and structure choice

| Structure | Access and order | Duplicates | Can change? | Useful choice |
| --- | --- | --- | --- | --- |
| List | Index/slice; insertion order | Yes | Yes | Changing ordered sequence group |
| Tuple | Index/slice; stored order | Yes | No item replacement | Fixed pair of lengths |
| Dictionary | Key lookup; insertion order | Keys unique; values may repeat | Yes | Identifier to DNA/properties |
| Set | Membership; no positional order | Unique hashable elements | Yes | Unique symbols/IDs |

`in` tests membership: list/tuple items, dictionary **keys**, set elements, string substring. `"AT" in dna_by_id` does not search dictionary values. Use `dna_by_id.values()` for that. `not in` reverses membership. Check before a lookup/removal when an item may be absent. Expect True, False, True below.

````python
dna_by_id = {"One": "AT", "Two": "GC"}
print("One" in dna_by_id)
print("AT" in dna_by_id)
print("AT" in dna_by_id.values())
````

Assignment can give two names the same mutable list. Changing it through one name affects what both see. A slice `[:]` copies this simple flat list so later item replacements do not affect the original. It is a shallow copy: nested mutable objects would still be shared. Expect ['TT','GC'], then ['AT','GC'], then ['TT','GC']. Practice 6: explain each output before running.

````python
original = ["AT", "GC"]
alias = original
alias[0] = "TT"
print(original)
original = ["AT", "GC"]
copy = original[:]
copy[0] = "TT"
print(original)
print(copy)
````

## 65-80: Iterate, then nest small records
Looping directly over a dictionary visits keys in insertion order. `.keys()` explicitly gives the keys; `.values()` gives the values. We use ordinary key lookup inside the loop to connect each ID with its DNA. Do not change dictionary size while iterating over it. Expect One AT, Two GC, then AT, GC.

````python
dna_by_id = {"One": "AT", "Two": "GC"}
for sequence_id in dna_by_id.keys():
    print(sequence_id, dna_by_id[sequence_id])
for sequence in dna_by_id.values():
    print(sequence)
````

A nested structure stores a collection inside another collection. First learn one record: properties map to values. Then put the record under its sequence identifier. Two steps are easier to read than a long chain.

```text
summaries
  "One" -> {"length": 2, "gc_percent": 0.0}
               "length" -> 2
```

`summaries["One"]` gets a dictionary record, then `record["length"]` gets its number. `summaries["One"]["length"]` performs both lookups. These synthetic properties describe AT only. Expect 2 then 0.0. Practice 7.

````python
record = {"length": 2, "gc_percent": 0.0}
summaries = {"One": record}
record = summaries["One"]
print(record["length"])
print(summaries["One"]["gc_percent"])
````

## 80-110: Analyze several named DNA strings
Problem: analyze four invented sequences without losing their identifiers. Inputs: a dictionary mapping unique labels to DNA. Outputs: a dictionary of summary records, readable per-sequence output, longest/shortest IDs and unique symbols. Identifiers are nonempty unique strings; these fixed examples satisfy that assumption. Repeated sequence values under different IDs are allowed.

Choose a dictionary for lookup by ID; choose another dictionary for each summary's named properties; choose a set for unique symbols. A list would be useful for repeated unnamed ordered strings, but dictionary keys avoid pairing IDs with positions manually.

Algorithm and pseudocode:
```text
CREATE empty summaries and unique-symbol set
SET longest and shortest IDs to empty text (nothing selected yet)
FOR each ID in input order:
    NORMALIZE its DNA
    RESET a four-key counts dictionary
    COUNT valid bases; mark invalid symbols
    IF empty or invalid: SHOW skipped ID and CONTINUE
    CALCULATE length and GC
    STORE summary under the same ID
    ADD its symbols to the set
    UPDATE longest only if first accepted or strictly longer
    UPDATE shortest only if first accepted or strictly shorter
DISPLAY each saved summary
IF no valid records: SHOW no valid sequences
ELSE: SHOW extremes and uniqueness
```

Build [02_DEMO.py](02_DEMO.py) in parts: input collection; outer ID loop; inner character loop with freshly reset counts; validation/GC; store summary; compare extremes; display. `counts[base] = counts[base] + 1` retrieves and updates a known key; `base in counts` checks a key before that access. `continue` skips invalid records. Empty strings never reach division.

Tie rule: keep the **first accepted identifier in input insertion order**. Use strict > for longest and < for shortest; equality must not replace an earlier selection. Empty ID text is a marker meaning no record selected, not a DNA record. The first accepted record initializes both extremes. With no accepted records, show No valid sequences. We do not rank invalid records.

Expected summaries:

| ID | DNA | Length | A | T | G | C | GC% |
| --- | --- | --- | --- | --- | --- | --- | --- |
| Demo_A | ATGCAT | 6 | 2 | 2 | 1 | 1 | about 33.33 |
| Demo_B | AT | 2 | 1 | 1 | 0 | 0 | 0.0 |
| Demo_C | GCGCAT | 6 | 1 | 1 | 2 | 2 | about 66.67 |
| Demo_D | ATAT | 4 | 2 | 2 | 0 | 0 | 0.0 |

Longest Demo_A 6 (ties Demo_C); shortest Demo_B 2. Four unique identifiers and four observed symbols. Output symbols in fixed ATGC order by membership checks because the set's own order is unspecified. The following complete repeat uses the script's same logic. Trace one outer iteration before Run All.

````python
# Synthetic educational DNA, invented for collection and arithmetic practice.
# Each identifier is unique. Dictionary insertion order sets the tie rule.
sequences = {"Demo_A": "ATGCAT", "Demo_B": "AT", "Demo_C": "GCGCAT", "Demo_D": "ATAT"}
summaries = {}
unique_symbols = set()
longest_id = ""
shortest_id = ""
longest_length = 0
shortest_length = 0

for sequence_id in sequences:
    sequence = sequences[sequence_id].strip().upper()
    counts = {"A": 0, "T": 0, "G": 0, "C": 0}
    valid_dna = True
    for base in sequence:
        if base in counts:
            counts[base] = counts[base] + 1
        else:
            valid_dna = False
    if len(sequence) == 0 or not valid_dna:
        print("Skipped invalid DNA:", sequence_id)
        continue
    sequence_length = len(sequence)
    gc_percent = (counts["G"] + counts["C"]) / sequence_length * 100
    summaries[sequence_id] = {"length": sequence_length, "A": counts["A"],
                              "T": counts["T"], "G": counts["G"], "C": counts["C"],
                              "gc_percent": gc_percent}
    for base in sequence:
        unique_symbols.add(base)
    # Strict comparison keeps the first accepted identifier on a length tie.
    if longest_id == "" or sequence_length > longest_length:
        longest_id = sequence_id
        longest_length = sequence_length
    if shortest_id == "" or sequence_length < shortest_length:
        shortest_id = sequence_id
        shortest_length = sequence_length

for sequence_id in summaries:
    record = summaries[sequence_id]
    print("Sequence:", sequence_id)
    print("Length:", record["length"])
    print("A:", record["A"], "T:", record["T"], "G:", record["G"], "C:", record["C"])
    print("GC percentage:", record["gc_percent"])

if len(summaries) == 0:
    print("No valid sequences.")
else:
    print("Longest:", longest_id, longest_length)
    print("Shortest:", shortest_id, shortest_length)
    print("Unique identifiers:", len(set(summaries)))
    # Check membership in a fixed order; set display order is not guaranteed.
    for base in "ATGC":
        if base in unique_symbols:
            print("Observed symbol:", base)
````
