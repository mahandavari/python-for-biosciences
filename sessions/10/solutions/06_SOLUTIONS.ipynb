For the one-column extension, calculate G counts for the already checked homework records. The small loop below prints the extended rows so you can compare them with your CSV change; it does not complete the final-project count/motif/chart report.


```python
with (course_root / "datasets" / "session-07" / "homework.fasta").open(encoding="utf-8") as handle:
    records = list(SeqIO.parse(handle, "fasta"))
print("record_id,length_bases,gc_percent,g_count")
for record in records:
    sequence_text = str(record.seq).strip().upper()
    g_count = 0
    c_count = 0
    for base in sequence_text:
        if base == "G":
            g_count = g_count + 1
        elif base == "C":
            c_count = c_count + 1
    gc_percent = (g_count + c_count) / len(sequence_text) * 100
    print(record.id, len(sequence_text), round(gc_percent, 2), g_count, sep=",")
```

Expected rows: Seed_A,4,50.0,1; Seed_B,4,0.0,0; Seed_C,6,100.0,3. Add the new header and value in the same column position in your copy of the guided script. Counts reset inside the loop.
