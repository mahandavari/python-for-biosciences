## 25-45: Comments, variables and assignment
A **comment** starts with `#`; Python ignores the rest of that line. Comments explain why we made a choice. A **variable** is a name referring to a value. In `sample_count = 3`, `=` assigns the value on the right to the name on the left. It is not a mathematical equation or a test of equality. Run this cell; expect 3.


````python
# This count is invented for practice.
sample_count = 3
print(sample_count)
````

Quotes mark text, also called a **string**. A name without quotes asks for the value assigned to it. `print(sample_count)` prints 3; `print("sample_count")` prints the name as text.

Use descriptive names such as `sample_count`. Names can use letters, digits and underscores, but cannot start with a digit. Use lowercase words separated by `_` (underscore). No spaces or hyphens. Names are case-sensitive: `sample_count` and `Sample_count` differ. Do not use reserved words such as `True` as names, or replace built-in names such as `print`. Common mistake: `sample count = 3` has a space and cannot be a variable name.

Assignment can replace a value. Predict the output, then run:


````python
sample_count = 3
sample_count = 4
print(sample_count)
````

Expected: 4. Python carries out lines in order; the last assignment changes the value the name refers to.

### Four basic types
An **int** is a whole number, such as 3. A **float** is a number with a decimal point, such as 12.5. A **str** is text in quotes. A **bool** is a truth value, `True` or `False`, with initial capitals and no quotes. Today we only store booleans; conditions come in Session 03. A decimal uses a dot, not a comma. `"3"` is text, not an integer.

`type(value)` tells us a value's type. It is a built-in operation, like print. The parentheses surround the value we give it. Here print displays what type returns. The word `class` in the output means the type; no class programming is needed.


````python
sample_count = 3
length_mm = 12.5
sample_name = "Demo A"
is_synthetic = True
````

````python
print(type(sample_count))
print(type(length_mm))
print(type(sample_name))
print(type(is_synthetic))
````

Expected lines: `<class 'int'>`, `<class 'float'>`, `<class 'str'>`, `<class 'bool'>`.

## 65-80: Arithmetic and evaluation order
Operators combine numbers. `+` adds; `-` subtracts; `*` multiplies; `/` divides; `//` does floor division (rounds the quotient down); `%` gives the remainder; `**` is a power. We use positive examples for floor division and remainder today. Ordinary `/` gives a float even when the answer is a whole number.

Predict the next outputs: 7, 3, 10, 2.5, 2, 1, 8.


````python
print(5 + 2)
print(5 - 2)
print(5 * 2)
print(5 / 2)
print(5 // 2)
print(5 % 2)
print(2 ** 3)
````

Parentheses group operations first. Powers come before multiplication/division/floor division/remainder, which come before addition/subtraction. Within the multiplication/division group or the addition/subtraction group, evaluate left to right. Keep expressions simple and use parentheses to make your intent clear.

Teacher demo: predict both lines below, then run. Expect **14**, then **20**.


````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

For an arithmetic mean, divide the entire sum by the number of values. All inputs must use the same unit. The three fictional seedling lengths from Session 01 now become a calculation, rather than just a typed message. Expected: `Mean length (mm): 12.0`.


````python
# Synthetic educational lengths in millimetres.
first_length_mm = 10.0
second_length_mm = 12.0
third_length_mm = 14.0
mean_length_mm = (first_length_mm + second_length_mm + third_length_mm) / 3
print("Mean length (mm):", mean_length_mm)
````

## 80-105: DNA counts and GC percentage
DNA stores biological information. We represent its four standard bases with letters: **A** (adenine), **T** (thymine), **G** (guanine), **C** (cytosine). A nucleotide is a building unit containing a base, sugar and phosphate. Here each count is the number of positions carrying one of those bases in one fictional sequence. Sequence length is the total number of positions. We assume only A, T, G and C are present.

**GC content** is the fraction of those positions that are G or C, often expressed as a percentage. We add G and C, divide by the total and multiply by 100. Counts do not reveal the order of bases, species, function or experimental effects.

### Problem, input and plan
Problem: compute a GC percentage from numeric counts. We are not counting a DNA string yet; loops and sequence processing come in Session 04. Inputs are invented counts: A=30, T=20, G=25, C=25. Units are bases (whole counts); the output percentage is a fraction multiplied by 100.

Assumptions: counts are **nonnegative whole numbers** and the **total is greater than zero**. With no bases there is no defined fraction to calculate; division by zero is invalid. The current script does not enforce these assumptions. Invalid text, negative counts and a zero total are not valid inputs. Input validation comes in Session 03.

Algorithm:
1. Store four counts.
2. Add all counts for the length.
3. Add G and C.
4. Divide GC count by length, then multiply by 100.
5. Print the length and percentage.

The formula is `(g_count + c_count) / (a_count + t_count + g_count + c_count) * 100`.

## 105-115: Repeat inside a notebook
Run the next cells in order to repeat the script. Each part shows one step. Expect total 100, GC count 50, then `GC percentage: 50.0`.


````python
# Synthetic educational counts; assume nonnegative counts and positive total.
a_count = 30
t_count = 20
g_count = 25
c_count = 25
````

````python
total_count = a_count + t_count + g_count + c_count
gc_count = g_count + c_count
print("Total bases:", total_count)
print("GC bases:", gc_count)
````

````python
gc_percent = gc_count / total_count * 100
print("GC percentage:", gc_percent)
````
