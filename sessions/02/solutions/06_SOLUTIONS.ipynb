## Practice 5: GC from counts
Assume four nonnegative whole counts with a positive total. Counts describe only A, T, G, C positions; they do not show sequence order. Expected length 80 bases and GC percentage 25.0. We define all inputs here so a new kernel can reproduce the result.


```python
# Synthetic educational counts, not experimental data.
a_count = 40
t_count = 20
g_count = 10
c_count = 10
total_count = a_count + t_count + g_count + c_count
gc_percent = (g_count + c_count) / total_count * 100
print("Total bases:", total_count)
print("GC percentage:", gc_percent)
```

The numerator is 20 GC bases out of 80 total bases, so 20 / 80 * 100 = 25.0%. A zero total has no defined GC percentage; formal input checking comes in Session 03.

## Homework notebook reference
Problem: find the arithmetic mean of three fictional seedling lengths. A seedling is a young plant. mm means millimetres. Values 10.0, 12.0, 14.0 were invented for arithmetic practice. Assume nonnegative numeric lengths in the same unit. No experimental or treatment conclusion follows from these invented values. Expected output: `Mean length (mm): 12.0`.


```python
# Synthetic educational lengths in millimetres.
first_length_mm = 10.0
second_length_mm = 12.0
third_length_mm = 14.0
mean_length_mm = (first_length_mm + second_length_mm + third_length_mm) / 3
print("Mean length (mm):", mean_length_mm)
```

Parentheses ensure the whole sum is divided by 3. Changing an early cell does not automatically rerun later cells. Restart Kernel, then Run All to clear old memory and verify cell order. The separate homework script uses float(input(...)) for three typed values; this fixed-value notebook is simpler to rerun.
