## 10-25: Comparisons give Boolean values
A bool is True or False, without quotes. `=` assigns; `==` asks whether two values are equal. `!=` means not equal; `>` greater than; `<` less than; `>=` greater than or equal; `<=` less than or equal. The equality part matters at a boundary. Predict the six outputs for 10 before running: True, False, False, False, True, True.

````python
reading = 10
print(reading == 10)
print(reading != 10)
print(reading > 10)
print(reading < 10)
print(reading >= 10)
print(reading <= 10)
````

`and` is True only when both tests are True. `or` is True if at least one is True. `not` reverses a Boolean. These examples use an invented length in mm (millimetres). The range 0-20 is a classroom rule. Comparisons run before and/or; use parentheses to make combinations readable. Expect True, False, False. Checkpoint: Practice 1-2; change the reading to 21 and predict all three values.

````python
reading = 10
print((reading >= 0) and (reading <= 20))
print((reading < 0) or (reading > 20))
print(not (reading == 10))
````

## 25-45: Choose a branch
`if test:` runs its indented lines when the test is True. `else:` runs when it is False. A colon opens a block. Use four spaces for each level; do not mix tabs and spaces. Lines at the same indentation belong to the same block. The unindented final print runs either way. Expect Positive, then Checked.

````python
reading = 10
if reading > 0:
    print("Positive")
else:
    print("Not positive")
print("Checked")
````

`elif` means test another case only if earlier tests were False. An if/elif/else chain selects at most one branch. Several separate if statements can select several branches. This range is artificial: below 10, 10 through 20 inclusive, above 20. At 10 the first test fails; the second succeeds. Expect Inside. Try Practice 3-4.

````python
reading = 10
if reading < 10:
    print("Below")
elif reading <= 20:
    print("Inside")
else:
    print("Above")
````

A nested if is inside another branch. Read the outer decision first. The inner lines have eight spaces. Here the inner if is reached only for nonnegative values; expect Positive valid. Checkpoint: Practice 5. Trace -1 and 0 without running first.

````python
reading = 10
if reading >= 0:
    if reading == 0:
        print("Zero")
    else:
        print("Positive valid")
else:
    print("Invalid")
````

## 75-105: Build the script in small parts
Open [02_DEMO.py](02_DEMO.py). First write the four `int(input(...))` lines. Next add the negative guard. Next nest the zero-total guard. The notebook repeat below uses fixed values and the same algorithm. Read it in the same four parts. Predict 40.0, middle before running. Then attempt Practice 7 in your own copy. Change only the four inputs, not the rules.

````python
# Synthetic educational counts; enter whole numbers, without units.
a_count = 3
t_count = 3
g_count = 2
c_count = 2

if a_count < 0 or t_count < 0 or g_count < 0 or c_count < 0:
    print("Invalid: counts must be nonnegative.")
else:
    total_count = a_count + t_count + g_count + c_count
    if total_count == 0:
        print("Invalid: total must be positive.")
    else:
        gc_percent = (g_count + c_count) / total_count * 100
        print("Total bases:", total_count)
        print("GC percentage:", gc_percent)
        # These bands are classroom rules, not biological standards.
        if gc_percent < 40:
            print("Teaching band: low")
        elif gc_percent <= 60:
            print("Teaching band: middle")
        else:
            print("Teaching band: high")
````

Self-check cases (A,T,G,C): (0,0,0,0) invalid total; (-1,1,1,1) invalid negative; (5,5,0,0) 0%, low; (0,0,5,5) 100%, high; (3,3,2,2) 40%, middle; (2,2,3,3) 60%, middle; (30,20,25,25) 50%, middle. These are numerical properties of invented counts only. Expected output labels match the script. A successful run with mixed counts does not test every branch.

Common mistakes: using `and` between the negative tests accepts some negatives; use or because any negative is invalid. `< 60` instead of `<= 60` puts 60 in the wrong band. Using `//` truncates the fraction. Separate if statements can print two bands. Printing a calculated value outside its valid branch may use an undefined variable.
