### Build validate_dna

Read its contract first. Trace the input, each decision and the returned result before moving to the next function.

```python
def validate_dna(sequence):
    """Accept nonempty DNA after trimming ends and converting to upper case."""
    sequence = sequence.strip().upper()
    if sequence == "":
        return False
    for base in sequence:
        if base not in "ATGC":
            return False
    return True
```

### Build count_nucleotides

Read its contract first. Trace the input, each decision and the returned result before moving to the next function.

```python
def count_nucleotides(sequence):
    """Return counts, or None for invalid/empty DNA."""
    sequence = sequence.strip().upper()
    if not validate_dna(sequence):
        return None
    counts = {"A": 0, "T": 0, "G": 0, "C": 0}
    for base in sequence:
        counts[base] = counts[base] + 1
    return counts
```

### Build calculate_gc_content

Read its contract first. Trace the input, each decision and the returned result before moving to the next function.

```python
def calculate_gc_content(sequence):
    """Return GC percent, or None for invalid/empty DNA."""
    sequence = sequence.strip().upper()
    counts = count_nucleotides(sequence)
    if counts is None:
        return None
    return (counts["G"] + counts["C"]) / len(sequence) * 100
```

### Build find_motif_positions

Read its contract first. Trace the input, each decision and the returned result before moving to the next function.

```python
def find_motif_positions(sequence, motif):
    """Return zero-based overlapping starts; None if either input is invalid."""
    sequence = sequence.strip().upper()
    motif = motif.strip().upper()
    if not validate_dna(sequence) or not validate_dna(motif):
        return None
    positions = []
    # Move by one so overlapping matches remain visible.
    for position in range(len(sequence) - len(motif) + 1):
        if sequence[position:position + len(motif)] == motif:
            positions.append(position)
    return positions
```

Compose the steps: returned values become print arguments or saved results. Expected: True, counts A3 T3 G1 C1, 25.0, [4], and [0, 2]. Python starts at index 0. The search visits every start by one, preserving overlaps. A match is text equality, with no functional claim. Task 5: change input to GCGC and motif GC; predict counts and positions.

```python
sequence = "ATGCATAT"
print(validate_dna(sequence))
print(count_nucleotides(sequence))
gc_percent = calculate_gc_content(sequence)
print(gc_percent)
print(find_motif_positions(sequence, "ATA"))
print(find_motif_positions("ATATAT", "ATA"))
```
