## 10-30: One character, then a string
A string is a sequence of characters, even if it has just one character. Quotes mark text. `len(sequence)` gives the number of characters. Brackets access a position: the first character has index 0. For length 6, the last index is 5. `sequence[6]` would raise IndexError. We do not run that broken line. Expect A, A, T, 6 below.

````python
base = "A"
sequence = "ATATAT"
print(base)
print(sequence[0])
print(sequence[5])
print(len(sequence))
````

A slice `sequence[start:stop]` takes characters from start up to, but not including, stop. `sequence[:3]` starts at 0; `sequence[3:]` continues to the end. Slicing past the end stops at the string boundary, unlike an out-of-range single index. Strings cannot be changed one character at a time by index; assign a new string instead.

```text
Python index    0 1 2 3 4 5
DNA character   A T A T A T
[0:3]           A T A         -> ATA
[2:5]               A T A     -> ATA
```
These are Python zero-based positions. A report using one-based biological coordinates would call the starts 1 and 3. We report 0 and 2 throughout this session. Expect ATA, TAT, AT below.

````python
sequence = "ATATAT"
print(sequence[:3])
print(sequence[3:])
print(sequence[4:99])
````

A method is an operation attached to a value: the dot in `text.upper()` selects it; parentheses run it. `upper()` returns uppercase text; `strip()` removes whitespace from the two ends, not from the middle. Store the returned value because the original string is unchanged. `count("A")` counts occurrences. First we learn this tool; later we build the counting algorithm manually. For longer substrings, count uses non-overlapping matches, so it is not our overlapping motif algorithm. Expect ATGC, 1, 4. Practice 1.

````python
raw_sequence = "  atGc  "
sequence = raw_sequence.strip().upper()
print(sequence)
print(sequence.count("A"))
print(len(sequence))
````

## 30-50: Repetition with for, while and range
`for base in sequence:` visits each character in order. The loop variable takes the current character. Its indented block repeats. A string is iterable: Python can visit its elements one by one. Here `in` belongs to the for syntax; as a separate test, `base in "ATGC"` checks membership. Expect A then T on separate lines.

````python
sequence = "AT"
for base in sequence:
    print(base)
````

`range(stop)` supplies integers 0 through stop-1. `range(start, stop)` starts at start and still excludes stop. It is useful when we need positions. A nonpositive stop with the default start 0 gives no iterations. Expect 0 A, 1 T.

````python
sequence = "AT"
for position in range(len(sequence)):
    print(position, sequence[position])
````

`while test:` repeats while the test stays True. It tests before every iteration, including the first. Update the counter or the loop may never stop. `position = position + 1` reads the old value, adds one, then stores the new value. For visits known characters directly; while is useful when a changing condition controls repetition. Expect positions 0,1,2. Trace this cell by writing the value before and after each update. Practice 2.

````python
position = 0
while position < 3:
    print(position)
    position = position + 1
````

## 50-65: Count one base, then four
Problem: find base counts from a string. Input ATGCAT; output initially A count 2. Plan: set a counter to zero before the loop, visit each character, add one only when it is A, print after the loop. Resetting the counter inside the loop loses earlier matches. A trace for the first three characters follows.

| Character | A count before | Is A? | A count after |
| --- | --- | --- | --- |
| A | 0 | True | 1 |
| T | 1 | False | 1 |
| G | 1 | False | 1 |

Run the small counter, then expand it to four counters with if/elif. No list or dictionary is needed yet.

````python
sequence = "ATGCAT"
a_count = 0
for base in sequence:
    if base == "A":
        a_count = a_count + 1
print("A:", a_count)
````

For this counting exercise assume only uppercase A/T/G/C. Initialize each counter once; each character can select only one base branch. Expect A:2, T:2, G:1, C:1. Practice 3. Later we reject strings that break the assumption.

````python
sequence = "ATGCAT"
a_count = 0
t_count = 0
g_count = 0
c_count = 0
for base in sequence:
    if base == "A":
        a_count = a_count + 1
    elif base == "T":
        t_count = t_count + 1
    elif base == "G":
        g_count = g_count + 1
    elif base == "C":
        c_count = c_count + 1
print("A:", a_count)
print("T:", t_count)
print("G:", g_count)
print("C:", c_count)
````

## 65-80: Search a short exact pattern
A motif here means a short DNA pattern, not a proven functional site. We compare exact text on the given strand only; no reverse complement or approximate matching. Input sequence ATATAT and motif ATA. Output positions 0 and 2, count 2. A count says how many; positions say where. Overlap means two matches share characters.

Algorithm: reject empty motif; set count to zero; try every possible start; take a slice as long as the motif; if equal, print the start and increment count; finally print count. The last legal start is length(sequence)-length(motif). Because range excludes stop, add 1 to include it. Move by one even after a match so overlaps remain visible.

```text
START  WINDOW  MATCH?  COUNT AFTER
0      ATA     Yes     1
1      TAT     No      1
2      ATA     Yes     2
3      TAT     No      2
```

The `window` variable simply stores the current slice. Empty motif is invalid by our course rule. A motif longer than the sequence gives no starts and count zero. For this first search, assume normalized A/T/G/C inputs. Expect Match position: 0, Match position: 2, Match count: 2.

````python
# Synthetic pattern search; fixed inputs make this cell repeatable.
sequence = "ATATAT".strip().upper()
motif = "ATA".strip().upper()
if len(motif) == 0:
    print("Invalid: empty motif.")
else:
    match_count = 0
    for position in range(len(sequence) - len(motif) + 1):
        window = sequence[position:position + len(motif)]
        if window == motif:
            print("Match position:", position)
            match_count = match_count + 1
    print("Match count:", match_count)
````

Checkpoint: Practice 4. Search AAA in AAAAA: starts 0,1,2. `sequence.count(motif)` would report 1 because it does not count overlaps. `range(len(sequence)-len(motif))` misses the last legal start. Do not jump by motif length after a match.

## 80-105: DNA analysis and GC from characters
Problem: replace manual numeric counts with a DNA string. Inputs are text strings for DNA and motif. Output: length, four counts, GC percentage and match positions/count, or clear rejection messages. Normalize both using strip then upper. Empty DNA is invalid. Accept only A/T/G/C; N, punctuation and internal spaces are invalid in this simplified alphabet. Real DNA text can contain ambiguity symbols, but they are outside today's accepted input.

Build [02_DEMO.py](02_DEMO.py) in parts: (1) read/normalize two strings; (2) initialize four counters and a valid_dna Boolean; (3) traverse characters, counting known bases and marking any other symbol invalid; (4) reject empty/invalid DNA; (5) in the valid branch, calculate `(G+C)/length*100`; (6) validate motif and add the exact search.

`base not in "ATGC"` means the character is not among those four letters. A Boolean flag remembers that an invalid symbol occurred; it stays False once invalid DNA is seen. We count while checking, but print no partial counts as a valid result. The checks occur before division. If only the motif is invalid, the valid DNA summary still prints, followed by a motif error and no match result.

Pseudocode:
```text
NORMALIZE sequence and motif
SET counts to zero and valid DNA to True
FOR each DNA character:
    INCREMENT the matching base counter, or mark DNA invalid
IF DNA empty or invalid: SHOW error
ELSE:
    SHOW length, counts and GC percentage
    CHECK motif characters and reject empty or invalid motif
    IF motif valid: SEARCH every legal start, SHOW positions and count
```
Run the script with `  atGc  ` then `gc`: length 4, each count 1, GC 50.0, position 2, count 1. The fixed-input notebook repeat below uses the same algorithm. Read one part at a time, then run. Practice 5-6.

````python
# Synthetic educational DNA only; no file or external dataset is used.
sequence = "  atGc  ".strip().upper()
motif = "gc".strip().upper()
valid_dna = True
a_count = 0
t_count = 0
g_count = 0
c_count = 0
for base in sequence:
    if base == "A":
        a_count = a_count + 1
    elif base == "T":
        t_count = t_count + 1
    elif base == "G":
        g_count = g_count + 1
    elif base == "C":
        c_count = c_count + 1
    else:
        valid_dna = False

if len(sequence) == 0:
    print("Invalid: empty DNA.")
elif not valid_dna:
    print("Invalid: DNA must contain only A, T, G, C.")
else:
    sequence_length = len(sequence)
    print("Length:", sequence_length)
    print("A:", a_count)
    print("T:", t_count)
    print("G:", g_count)
    print("C:", c_count)
    gc_percent = (g_count + c_count) / sequence_length * 100
    print("GC percentage:", gc_percent)
    valid_motif = True
    for base in motif:
        if base not in "ATGC":
            valid_motif = False
    if len(motif) == 0:
        print("Invalid: empty motif.")
    elif not valid_motif:
        print("Invalid: motif must contain only A, T, G, C.")
    else:
        match_count = 0
        # Include the last legal start; moving by one preserves overlaps.
        for position in range(len(sequence) - len(motif) + 1):
            if sequence[position:position + len(motif)] == motif:
                print("Match position:", position)
                match_count = match_count + 1
        print("Match count:", match_count)
````

After understanding the loop, compare the shorter method below for known-valid ATGC. Both get GC 50.0. `count("G") + count("C")` counts the same bases as the explicit loop; it does not replace the need to validate the alphabet or guard empty strings. A frequency is count divided by total: here each base frequency is 0.25 (25%). The four frequencies sum to 1.

````python
sequence = "ATGC"
print("GC percentage:", (sequence.count("G") + sequence.count("C")) / len(sequence) * 100)
print("A frequency:", sequence.count("A") / len(sequence))
````

## 105-115: Stop or skip, then check edge cases
`break` ends the nearest loop immediately. `continue` skips the remaining lines of the current iteration and moves to the next iteration. The first cell prints A,T, then stops at N. The second prints A,T,G,C by skipping N. This illustrates control flow only: silently removing N changes the data, so our actual analyzer rejects invalid DNA instead. Practice 7.

````python
for base in "ATNGC":
    if base == "N":
        break
    print(base)
print("Skip demonstration:")
for base in "ATNGC":
    if base == "N":
        continue
    print(base)
````

Self-check the script: empty/whitespace DNA -> Invalid: empty DNA.; ATN or AT GC -> invalid DNA; ATATAT/ATA -> positions 0,2, count 2, GC 0%; ATGC/TT -> count 0; AT/ATG -> count 0; ATGC/empty motif -> motif error after valid summary; ATGC/GN -> invalid motif; ATGC/ATGC -> position 0, count 1. Mixed case is normalized. A long motif is valid but absent, not malformed. In the isolated search cell, an empty sequence with a nonempty motif gives count 0; the full analyzer rejects empty DNA because GC is undefined.

Common mistakes: omitting +1 loses a final match; resetting counts in a loop loses history; printing the summary inside the traversal produces unfinished totals; using a list of positions would require Session 05, so today print positions individually. Motif presence does not prove biological significance.
