## Counting and frequencies
ATGCAT gives A2,T2,G1,C1, length 6, GC about 33.33%. We assume valid nonempty uppercase DNA for this focused cell; the full homework cell below validates. Set counters before the loop.

```python
# Synthetic educational DNA string; no biological meaning is inferred.
sequence = "ATGCAT"
for base in sequence:
    print(base)
a_count = 0
t_count = 0
g_count = 0
c_count = 0
for base in sequence:
    if base == "A":
        a_count = a_count + 1
    elif base == "T":
        t_count = t_count + 1
    elif base == "G":
        g_count = g_count + 1
    elif base == "C":
        c_count = c_count + 1
print("A:", a_count)
print("T:", t_count)
print("G:", g_count)
print("C:", c_count)
if len(sequence) > 0:
    print("GC percentage:", (g_count + c_count) / len(sequence) * 100)
```

Frequency is count/length. Four frequencies add to approximately 1 (floating-point approximations may affect the last digit).

```python
print("A frequency:", a_count / len(sequence))
print("T frequency:", t_count / len(sequence))
print("G frequency:", g_count / len(sequence))
print("C frequency:", c_count / len(sequence))
print("Frequency sum:", (a_count + t_count + g_count + c_count) / len(sequence))
```

## Exact overlapping search
ATATAT/ATA gives starts 0,2 and count 2. The empty motif guard prevents a meaningless match at every boundary; one-position steps keep overlaps.

```python
# Synthetic pattern search; fixed inputs make this cell repeatable.
sequence = "ATATAT".strip().upper()
motif = "ATA".strip().upper()
if len(motif) == 0:
    print("Invalid: empty motif.")
else:
    match_count = 0
    for position in range(len(sequence) - len(motif) + 1):
        window = sequence[position:position + len(motif)]
        if window == motif:
            print("Match position:", position)
            match_count = match_count + 1
    print("Match count:", match_count)
```

## Complete homework fixed-input reference
Input atgcATAT/ata normalizes to ATGCATAT/ATA. Expected length 8, A3,T3,G1,C1, GC 25.0, match at 4, count 1. Invalid DNA suppresses calculation; invalid motif suppresses only the search. See the terminal script for interactive input.

```python
# Synthetic educational DNA only; no file or external dataset is used.
sequence = "atgcATAT".strip().upper()
motif = "ata".strip().upper()
valid_dna = True
a_count = 0
t_count = 0
g_count = 0
c_count = 0
for base in sequence:
    if base == "A":
        a_count = a_count + 1
    elif base == "T":
        t_count = t_count + 1
    elif base == "G":
        g_count = g_count + 1
    elif base == "C":
        c_count = c_count + 1
    else:
        valid_dna = False

if len(sequence) == 0:
    print("Invalid: empty DNA.")
elif not valid_dna:
    print("Invalid: DNA must contain only A, T, G, C.")
else:
    sequence_length = len(sequence)
    print("Length:", sequence_length)
    print("A:", a_count)
    print("T:", t_count)
    print("G:", g_count)
    print("C:", c_count)
    gc_percent = (g_count + c_count) / sequence_length * 100
    print("GC percentage:", gc_percent)
    valid_motif = True
    for base in motif:
        if base not in "ATGC":
            valid_motif = False
    if len(motif) == 0:
        print("Invalid: empty motif.")
    elif not valid_motif:
        print("Invalid: motif must contain only A, T, G, C.")
    else:
        match_count = 0
        # Include the last legal start; moving by one preserves overlaps.
        for position in range(len(sequence) - len(motif) + 1):
            if sequence[position:position + len(motif)] == motif:
                print("Match position:", position)
                match_count = match_count + 1
        print("Match count:", match_count)
```
